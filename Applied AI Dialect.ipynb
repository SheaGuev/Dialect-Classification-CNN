# Literature Review: Speech and Text Recognition

The utlisation of Artificial Intelligence in Speech and Text, has been integral for both accessibility and ease of use for the day to day lives of individuals. From AI Assistants to converting text, language translation and general interactivity of users with software, the use of speech and text recognition positively impacts multiple industries such as smartphones, personal assistants, language translation, security, and even forensic analysis.

## AI in Speech & Text Recognition

Speech recognition is the conversion of speech in to text in some sense to understand what a person is saying, it can also extend to translation from one spoken language to another. A common use for this is in text-to-speech, and with personal assistants Before the development of ASR, natural language processing with machine learning started with N-grams which could predict the probability of a sentence based on the number of syllables, and has now developed into much more complex understanding in human speech.

When it comes to text recognition it is quite different to speech, as it requires the extraction of text from images, which is image processing instead of audio processing. The commonality between text and speech is the accessibility they provide, being able to communicate with machines without directly typing in text can make computers easier to use for some.

### How AI was Applied:

**Speech Recognition**  
**Automatic Speech Recognition (ASR):** Automatic speech recognition has integrated AI into the understanding and conversion of human speech into text, this has been vital for personal assistants, language translation and more. (Bhavani and Kathirvelu, 2024), (Swaroop Reddy Gayam, 2021\)

Through the development of techniques like Convolutional Neural Networks (CNNs) we were able to capture details of acoustic features, and then by implements Recurrent Neural Networks (RNNs) which are Used for modelling sequential dependencies within speech patterns \[3\]. Other examples of deep neural networks now used include Long-Short-Term-Memory, and transformer architectures have been able to improve recognition accuracy as they are able to analyse the temporal differences over time in how sound changes at a deeper level.

**Text-To-Speech:** Text-To-Speech is the generation of speech from text, which again is done with a form of deep learning. TTS is a two step process which requires an encoder, coupled with the generative model. The model will produce a variation of frequencies like a spectogram, which is then converted into audio by the encoder. 

There are several models available which can synthesise realistic speech from text, including WaveNet which is a very notable generative model created by DeepMind. WaveNet utilises deep convolutional neural networks to generate the speech from the text given dependant on the sample, this has made TTS more dynamic and fit to a dynamic range of voice types. In the creation of speech synthesis with deep neural networks, multiple models work together to create the final output, for example one model may be for the acoustic, one for the duration, and one for the pitch, but by combing all 3 models we can ge the final output of how the spectogram.

**Natural Language Understanding (NLU):** Natural language understanding is the ability for computers to understand meaning of text by comparing contextual data which can be used to in form decision making processes, or in interaction with a large language model in chatbot form like ChatGPT(Matic et al., 2021\). 

This has developed by training models on context, and has been helped by the use of embeddings which have allowed text to be translate into vectors and other mathematical representations. By representing text as vectors its easier to feed more context into these models and therefore producing better overall results. There are also privacy centred solutions like BERT(Qu et al., 2021\) which is centred around privacy but while integrating as much data into the model as possible through chatbots.

**Audio-Visual Speech Recognition (AVSR):** AVSR combines visual cues for speech like lip reading and body language, with ASR to more accurately predict what a person is saying. This could prove very helpful in noisy environments and is promisingly integrating.

An example of how AI has been integrated is through transformer models with CTC Loss, which is used to train RNNs on problems like speech where the timing is variable. Also it has been found that by adding noise in the training (Bharath et al., 2024\), it can cause predictions to be more accurate during real world examples.

## Summary

The challenges that arise during all these technologies are varied and complex. These problems are at the cusp of what has been created, but are actively being improved with great speed. 

The improvement on speech and text processing with AI has drastically improved with generative models like text to speech, understanding with NLU and Speech Recognition, as well as combing different models for greater accuracy. There is still a lot of exciting work to be done in this area of development, but through advanced deep learning techniques, it’s advancement is rapid.

# B: Dialect Classification from Speech:

The 3 techniques I’ve chosen for this are…

Convolution Neural Networks, Recurrent Neural Networks, and Transformers. All of these options belong to the subject of deep learning, but all have different strengths and weaknesses. 

**CNNs** were chosen as they are great at extracting spatial patterns like spectrograms (used for audio processing) and can recognise similar patterns in the pathways of sounds, they are not too computationally expensive and have been shown to work well in extracting audio features (Alansari, 2023\). 

**CNNs Strengths:**

* Work well at capturing patterns in spectrogram representations of speech  
* Can extract spatial features from speech such as phonetic patterns   
* Can easily be combined with other methods to cover spatial and temporal patterns

**CNNs Weaknesses:** 

* Limited ability to model long-term temporal differences over time which can be important for extracting features and patterns from sound  
* It is computationally expensive in comparison to previous ml models like Support Vector Machines (SVM)

**RNNs** are fantastic at temporal sequence modelling (Song, Thi and Ta, 2023\), which can track changes over time and find differences in data over time steps. This makes them very useful for something like audio which is changing at each frequency.

**RNNs Strengths:**

* Works very well for speech as it is sequential data which changes with each time step, these changes per time step can give insight into the dynamics patterns of dialects  
* Its been found that using bidirectional LTSMs (Alansari, 2023\) improve performance of the RNN by taking into account both past and future context

**RNNs Weaknesses:** 

* RNNs can struggle with longer sequences due to vanishing gradient problems (can be overcome using LTSM)  
* It is computationally expensive and can take much longer to train than tradition ml models

**Transformer models** are able to capture long range dependencies through attention mechanisms (Baniata and Kang, 2023\), this could make them effective in extracting features from audio over time, it also true that pre-trained transformers are available which can be fine tuned on smaller dataset, which may make training less computationally expensive and require less data.

**Transformers Strengths:**

* Gives high level performance in text classification due to its ability to capture macro/long-range patterns (Humayun et al., 2022\)  
* Pre-trained transformers can be fine tuned saving time and computational power  
* Effective for multiple modalities like speech and text  
* Great for text analysis as can differ between words used between dialects which could save in depth analysis of speech in some cases   
* If use in conjunction with a CNN/RNN it may be able to pick up on macro details like differences in words used between dialects easier to contribute to the model

**Transformers Weaknesses:** 

* Less useful for speech as it is text analysis   
* If used in conjunction with a CNN/RNN it may not provide much more use than tradition ml methods for the increase in computation  
* May overfit easily on smaller datasets 

**Conclusion:**  
Overall there are multiple approaches to the problem of dialect classication which are able to utilise different aspects of patters in speech. The most complex but also likely best performing would be a deepm neural network which incorporates all 3 of these approaches to capture as much information as possible for the training, but this may be time consuming and computationally expensive. Being able to combing temporal, macro, and spatial features of speech can be powerful, but for the sake of both computational efficiency, and deeper exploration I opt to utilise a CNN due to its cohesiveness with spectrographic analysis, aswell as its ability to detect phonetic patters. Although somewhat computationally expensive, it has been shown to produce some high accuracy in comparison to traditional methods like SVM, implementing this should give deeper insight into speech analysis.

```python

import librosa


# Function to generate spectrogram
def generate_spectrogram(audio_path):
    try:
        # Load audio file
        y, sr = librosa.load(audio_path, sr=None)
        
        # Generate Mel spectrogram
        # mel_spec = librosa.feature.melspectrogram(y=y, sr=sr, n_fft=2048, hop_length=512, n_mels=128)
        mel_spec = librosa.feature.melspectrogram(y=y, sr=sr, n_mels=64) 
        
        # Convert to decibel scale
        mel_spec_db = librosa.power_to_db(mel_spec, ref=np.max)


        #3. Apply Per-Channel Energy Normalization (PCEN)

        pcen_S = librosa.pcen(mel_spec * (2**31), sr=sr, hop_length=512)

        # mel_spec_db = mel_spec_db[..., np.newaxis]##Adding a 3rd axis to make it 3d so it fits in the CNN

        
        # return [mel_spec_db, sr]
        return [pcen_S, sr]

    except Exception as e:
        print(f"Error processing {audio_path}: {e}")
        return None


# Iterate through the DataFrame rows and add spectrograms
spectrograms = []
srs = []
for index, row in train_1_df.iterrows():

    
    audio_path = os.path.join(data_folder, row['path'])
    

    ##TRBLSHOOTING Errors with literal string escapes
    # audio_path = f"{data_path}\\{audio_path}"
    audio_path = audio_path.replace("\\\\", "\\")
    # print(audio_path)
    
    spectrogram = generate_spectrogram(str(audio_path))  # Generate spectrogram for the audio path
    spectrograms.append(spectrogram[0])  # Append the spectrogram to the list
    srs.append(spectrogram[1])
    
print(len(spectrograms))


# matching_rows_specs = []
# matching_srs = []
# for index, row in matching_rows.iterrows():
#     audio_path = os.path.join(data_folder, row['path_from_data_dir_windows'])
#     audio_path = audio_path.replace("\\\\", "\\")
    
#     spectrogram = generate_spectrogram(str(audio_path))  # Generate spectrogram for the audio path
#     matching_rows_specs.append(spectrogram[0])  # Append the spectrogram to the list
#     matching_srs.append(spectrogram[1])

## NOTE: THE ABOVE WAS USED TO COMPARE SPECTROGRAMS OF MATCHING FILES TO VERIFY THEY WERE DUPLICATES
```

```python
def fix_spectrogram_length(mel_spec_db, target_length):
    current_length = mel_spec_db.shape[1]
    
    if current_length < target_length:
        # Pad with zeros along the time axis
        padding = target_length - current_length
        mel_spec_db = np.pad(mel_spec_db, ((0, 0), (0, padding)), mode='constant')

    elif current_length > target_length:
        # Trim to the target length
        mel_spec_db = mel_spec_db[:, :target_length]
    
    return mel_spec_db

# target_length = 245  # Example fixed length for all spectrograms

target_length = 144

fixed_spectrograms = [fix_spectrogram_length(spec, target_length) for spec in spectrograms]
print(fixed_spectrograms[1].shape, spectrograms[1].shape)
```

```python
##CREATE THE CNN 

import tensorflow as tf
from tensorflow.keras import layers, models

def create_cnn(input_shape=(64, 144, 1), num_classes=8):
    model = models.Sequential()

    # Convolutional block 1
    model.add(layers.Conv2D(filters=32, kernel_size=(3, 3), 
                            activation='relu', padding='same', 
                            input_shape=input_shape))
    
    model.add(layers.BatchNormalization())
    model.add(layers.MaxPooling2D(pool_size=(2, 2)))
    model.add(layers.Dropout(0.16)) ## Trying to reduce the overfitting

    # Convolutional block 2
    model.add(layers.Conv2D(filters=64, kernel_size=(3, 3), 
                            activation='relu', padding='same'))
    model.add(layers.BatchNormalization())
    model.add(layers.MaxPooling2D(pool_size=(2, 2)))

    model.add(layers.Dropout(0.24)) ## Trying to reduce the overfitting


    # Convolutional block 3
    model.add(layers.Conv2D(filters=128, kernel_size=(3, 3), 
                            activation='relu', padding='same'))
    model.add(layers.BatchNormalization())
    model.add(layers.MaxPooling2D(pool_size=(2, 2)))
    model.add(layers.Dropout(0.16))

    # Flatten and fully connected layers
    model.add(layers.Flatten())
    model.add(layers.Dense(128, activation='relu'))
    model.add(layers.BatchNormalization())
    model.add(layers.Dropout(0.08))  # Helps reduce overfitting
    model.add(layers.Dense(num_classes, activation='softmax'))

    return model

# Create the model
model = create_cnn(input_shape=(64, 144, 1), num_classes=8)

model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=1e-4),
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])
```
